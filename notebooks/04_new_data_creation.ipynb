```python
import sqlite3
import pandas as pd
from pathlib import Path

# Move from notebooks/ to the project root
project_root = Path.cwd().parent

# Database location
database_path = project_root / "data" / "product_analytics.db"

# Make sure the data folder exists
database_path.parent.mkdir(parents=True, exist_ok=True)

print(database_path)
```

```python
with sqlite3.connect(database_path) as connection:

    session_df.to_sql(
        "session_summary",
        connection,
        if_exists="replace",
        index=False
    )

    experiment_priority.to_sql(
        "experiment_priority",
        connection,
        if_exists="replace",
        index=False
    )

print("Database created successfully.")
```

```python
with sqlite3.connect(database_path) as connection:

    tables = pd.read_sql_query(
        """
        SELECT name
        FROM sqlite_master
        WHERE type = 'table'
        ORDER BY name;
        """,
        connection
    )

tables
```

```python
with sqlite3.connect(database_path) as connection:

    session_count = pd.read_sql_query(
        """
        SELECT COUNT(*) AS number_of_sessions
        FROM session_summary;
        """,
        connection
    )

session_count
```
