# Experiment Prioritization

## Business Objective

The previous analyses identified several product characteristics associated with customer engagement.

The objective of this notebook is to prioritize future product experiments based on statistical evidence and business relevance.

Rather than proving causality, this notebook translates observational analyses into actionable A/B testing hypotheses.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
summary = pd.DataFrame({

    "Business Question":[
        "First Product Category",
        "First Product Position",
        "First Product Photo",
        "First Product Price"
    ],

    "Statistical Test":[
        "Kruskal-Wallis",
        "Kruskal-Wallis",
        "Mann-Whitney U",
        "Mann-Whitney U"
    ],

    "P-value":[
        "<0.001",
        "<0.001",
        "<0.001",
        "<0.001"
    ],

    "Effect Size":[
        "ε² = 0.0125",
        "ε² = 0.0093",
        "r = 0.0647",
        "r = -0.0311"
    ],

    "Main Finding":[
        "Trousers received the highest engagement.",
        "Top Left position consistently received the highest engagement.",
        "Very small difference between photo types.",
        "Very small difference between price categories."
    ]

})

summary
```

```python
priority = pd.DataFrame({
    "Experiment": [
        "Homepage Product Position",
        "Homepage Category Ordering",
        "Product Photo Type",
        "Product Price Category"
    ],
    "Priority Score": [5, 4, 2, 1]
})

priority = priority.sort_values("Priority Score")

plt.figure(figsize=(9,5))
plt.barh(priority["Experiment"], priority["Priority Score"])
plt.xlabel("Priority Score")
plt.title("Recommended Product Experiment Priorities")
plt.tight_layout()
plt.show()
```

## Executive Recommendation

The analyses identified statistically significant associations between several product characteristics and customer engagement.

Among the evaluated features, product position and product category showed the strongest practical potential for future experimentation.

Although photo type and price category also produced statistically significant results, their practical effects were negligible. These should therefore receive lower priority.

Based on the current evidence, the recommended order for future A/B testing is:

1. Product Position
2. Product Category
3. Product Photo Type
4. Product Price Category

Because these findings are based on observational data, randomized controlled experiments are required before implementing product changes.

## Limitations

- The analyses are based on observational clickstream data.
- Statistical significance does not imply causality.
- Product characteristics may be correlated with other unmeasured factors.
- The proposed experiments are hypotheses that require validation through randomized A/B testing.
